```python
##Challenge 1
def is_balanced(s: str) -> bool:
    stack = []

    pairs = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for char in s:
        if char in '([{':
            stack.append(char)

        elif char in ')]}':
            if not stack or stack.pop() != pairs[char]:
                return False

    return len(stack) == 0


##Challenge 2

def two_sum(nums: list[int], target: int) -> tuple[int, int]:
    seen = {}  #
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i



am = [1,2,3,4,9,10]
print(two_sum(am, 7))

##Challenge 3

class Node:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next


def reverse_list(head: Node | None) -> Node | None:
    prev = None
    curr = head

    while curr is not None:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node

    return prev


n3 = Node(3)
n2 = Node(2, n3)
n1 = Node(1, n2)

new_head = reverse_list(n1)
curr = new_head
while curr is not None:
    print(curr.val, end=" ")
    curr = curr.next


##Challenge 8

def is_valid_bst(root: Node | None) -> bool:
    # Assumption: BST values must be distinct.
    def check(node, low, high):
        if node is None:
            return True

        # Current value must be strictly between its bounds.
        if not (low < node.val < high):
            return False

        # Every node in the left subtree must be smaller.
        # Every node in the right subtree must be larger.
        return (
            check(node.left, low, node.val)
            and check(node.right, node.val, high)
        )

    return check(root, float("-inf"), float("inf"))


#Challenge 4
class QueueFromStacks:
    def __init__(self):
        self.in_stack = []
        self.out_stack = []

    def enqueue(self, x) -> None:
        self.in_stack.append(x)

    def dequeue(self):
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())
        if not self.out_stack:
            raise IndexError("dequeue from empty queue")
        return self.out_stack.pop()


q = QueueFromStacks()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
print(q.dequeue())
print(q.dequeue())
q.enqueue(4)
print(q.dequeue())
print(q.dequeue())
```
